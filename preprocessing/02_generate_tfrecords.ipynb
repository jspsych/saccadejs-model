# Generate .tfrecords from public jpg data

This notebook will download the public JPG dataset from the OSF and generate the three different tfrecords datasets:

- Raw JPGs
- Landmarks only
- Landmarks and eye images

To authenticate to the OSF, you will need to add secrets. This is a Colab feature and can be done on the left side (key icon). Add a secret called *osfusername* with your OSF email login and a secret called *osftoken* that is a personal access token for the OSF that has write privlidges for the EyeSeeVC OSF project.

For the upload to OSF step to work, you'll need to delete the existing .tfrecords files in the OSF project.
